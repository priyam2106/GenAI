```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    
)
```

### Discuss the problem

```python
text = "hello I am Priyam Raj the Prime Minister of India for 15 years,"
```

```python
res = llm.invoke(f"Please Give me only name of Prime Minister and from How many years he was Prime minister: {text}")
```

```python
print(res.content)
```

### Solution is : Srectured O/P

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStrecture(BaseModel):
    name:str = Field(description="Complete Name")
    Post:str = Field(description="Post")
    Working_Year:int = Field(description="Total Years")

strectured_llm = llm.with_structured_output(ResponseStrecture)

```

```python
res = strectured_llm.invoke(f"Please Give me only name of Prime Minister and from How many years he was Prime minister: {text}")
```
