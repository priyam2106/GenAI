### Splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text="Today , I want to go to the school for only 1st half"\
"But i get fever i the early morning so i am unable to the school"
```

```python
splitter =  RecursiveCharacterTextSplitter(chunk_size=30,chunk_overlap=10)
```

```python
chunks=splitter.split_text(text)
```

```python
chunk = splitter.split_documents(texts)
```

### Vector Embeddingd and Vector Database

```python
documents = [
    "Machine learning is subset of ai",
    "deep learning uses neural network",
    "Python is the programming language",
    "India won the cricket world cup"
]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_groq import ChatGroq

from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully!")
```

```python
vector = embeddings.embed_documents(documents)
```
